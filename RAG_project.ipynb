```python

import fitz

# Open the PDF
doc = fitz.open(pdf_name)

# Create an empty list to store the text
pages_text = []

# Read each page
for page in doc:
    text = page.get_text()
    pages_text.append(text)

# Display results
print("Number of pages:", len(pages_text))
print("\nFirst page text:")
print(pages_text[0][:1000])
```

```python

# Create an empty list for cleaned pages
cleaned_pages = []

# Go through each page
for text in pages_text:

    # Split text into words and join with single spaces
    cleaned_text = " ".join(text.split())

    # Save the cleaned text
    cleaned_pages.append(cleaned_text)

# Show the first 500 characters
print(cleaned_pages[0][:500])
```

```python

# Define chunk settings
chunk_size = 100
overlap = 20

# Create an empty list to store chunks
chunks = []

# Go through each page
for page_number, text in enumerate(cleaned_pages):

    # Split the page into words
    words = text.split()

    # Move through the words in steps
    for start in range(0, len(words), chunk_size - overlap):

        # Select words for this chunk
        chunk_words = words[start:start + chunk_size]

        # Combine words into text
        chunk_text = " ".join(chunk_words)

        # Save the chunk and its page number
        if chunk_text:
            chunks.append({
                "text": chunk_text,
                "page": page_number + 1
            })

# Display results
print("Total chunks:", len(chunks))
print("\nFirst chunk:")
print(chunks[0])
```

```python

import numpy as np
import torch

def ask_rag(question, top_k=5):

    # STEP 1: Convert the question into a vector
    question_embedding = model.encode(
        [question],
        normalize_embeddings=True
    )

    query_vector = np.asarray(
        question_embedding,
        dtype="float32"
    )

    # STEP 2: Search FAISS for relevant chunks
    k = min(top_k, index.ntotal)

    scores, indices = index.search(query_vector, k)

    # STEP 3: Build context from retrieved chunks
    context_parts = []
    sources = []

    for chunk_index in indices[0]:

        if chunk_index == -1:
            continue

        chunk = chunks[int(chunk_index)]

        context_parts.append(
            f"[Page {chunk['page']}] {chunk['text']}"
        )

        sources.append(chunk["page"])

    context = "\n\n".join(context_parts)

    # STEP 4: Build a strict RAG prompt
    prompt = f"""
You are a research assistant answering questions about a PDF.

Rules:
1. Answer using ONLY the provided context.
2. Do not add facts that are not in the context.
3. Cite the page number for each factual claim.
4. If the answer is not supported by the context, say:
   "I cannot find the answer in the provided document."
5. Do not confuse topics mentioned with main findings.
6. Keep the answer concise and clear.

Context:
{context}

Question:
{question}

Answer:
"""

    # STEP 5: Prepare the prompt for Qwen
    messages = [
        {"role": "user", "content": prompt}
    ]

    formatted_prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        formatted_prompt,
        return_tensors="pt",
        return_attention_mask=True
    ).to(llm.device)

    # STEP 6: Generate the answer
    with torch.no_grad():
        outputs = llm.generate(
            **inputs,
            max_new_tokens=200,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    new_tokens = outputs[0][inputs["input_ids"].shape[1]:]

    answer = tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    )

    # STEP 7: Return the answer and source pages
    return {
        "question": question,
        "answer": answer,
        "source_pages": sorted(set(sources))
    }
```

```python

# Define our five evaluation questions
evaluation_questions = [
    "How does effective stress affect permeability?",
    "What is the stress sensitivity coefficient alpha?",
    "How does stress sensitivity affect oil production?",
    "What role do fractures play in the reservoir model?",
    "What are the main findings of this study?"
]

# Rebuild our ground-truth dictionary
ground_truth = {
    evaluation_questions[0]: [28, 6, 27],
    evaluation_questions[1]: [28, 32],
    evaluation_questions[2]: [65, 61, 66, 6],
    evaluation_questions[3]: [29, 25, 4],
    evaluation_questions[4]: [77, 78, 79, 80]
}

# Display the results
for i, question in enumerate(evaluation_questions, start=1):
    print(f"Question {i}: {question}")
    print("Relevant chunks:", ground_truth[question])
    print()
```

```python

import numpy as np

def evaluate_retrieval(questions, ground_truth, top_k=5):

    results = []

    for question in questions:

        # Step 1: Convert question into an embedding
        question_embedding = model.encode(
            [question],
            normalize_embeddings=True
        )

        query_vector = np.asarray(
            question_embedding,
            dtype="float32"
        )

        # Step 2: Retrieve chunks from FAISS
        scores, indices = index.search(
            query_vector,
            min(top_k, index.ntotal)
        )

        retrieved_ids = [
            int(i) for i in indices[0] if i != -1
        ]

        # Step 3: Get manually labeled relevant chunks
        relevant_ids = set(ground_truth[question])

        # Step 4: Calculate Hit@K
        hits = [
            i for i in retrieved_ids if i in relevant_ids
        ]

        hit_at_k = int(len(hits) > 0)

        # Step 5: Calculate Recall@K
        recall_at_k = (
            len(hits) / len(relevant_ids)
            if relevant_ids else 0
        )

        # Step 6: Calculate reciprocal rank
        reciprocal_rank = 0

        for rank, chunk_id in enumerate(
            retrieved_ids, start=1
        ):
            if chunk_id in relevant_ids:
                reciprocal_rank = 1 / rank
                break

        # Step 7: Save results
        results.append({
            "question": question,
            "retrieved_ids": retrieved_ids,
            "hit_at_k": hit_at_k,
            "recall_at_k": recall_at_k,
            "reciprocal_rank": reciprocal_rank
        })

    return results
```

```python

results = evaluate_retrieval(
    evaluation_questions,
    ground_truth,
    top_k=5
)

for i, result in enumerate(results, start=1):

    print("=" * 60)
    print("Question:", i)
    print(result["question"])

    print("Retrieved:", result["retrieved_ids"])
    print("Hit@5:", result["hit_at_k"])
    print("Recall@5:", round(result["recall_at_k"], 3))
    print("Reciprocal Rank:", result["reciprocal_rank"])
```

```python

average_hit = np.mean([
    r["hit_at_k"] for r in results
])

average_recall = np.mean([
    r["recall_at_k"] for r in results
])

mrr = np.mean([
    r["reciprocal_rank"] for r in results
])

print("FINAL RETRIEVAL METRICS")
print("-----------------------")
print("Hit@5:", round(average_hit, 3))
print("Recall@5:", round(average_recall, 3))
print("MRR@5:", round(mrr, 3))
```

```python

def evaluate_rewritten_retrieval(
    questions,
    ground_truth,
    rewritten_queries,
    top_k=5
):

    results = []

    for question in questions:

        # Use the rewritten query for searching
        search_query = rewritten_queries[question]

        query_embedding = model.encode(
            [search_query],
            normalize_embeddings=True
        )

        query_vector = np.asarray(
            query_embedding,
            dtype="float32"
        )

        scores, indices = index.search(
            query_vector,
            min(top_k, index.ntotal)
        )

        retrieved_ids = [
            int(i) for i in indices[0] if i != -1
        ]

        relevant_ids = set(ground_truth[question])

        hits = [
            i for i in retrieved_ids
            if i in relevant_ids
        ]

        hit_at_k = int(len(hits) > 0)

        recall_at_k = (
            len(hits) / len(relevant_ids)
            if relevant_ids else 0
        )

        reciprocal_rank = 0

        for rank, chunk_id in enumerate(
            retrieved_ids, start=1
        ):
            if chunk_id in relevant_ids:
                reciprocal_rank = 1 / rank
                break

        results.append({
            "question": question,
            "retrieved_ids": retrieved_ids,
            "hit_at_k": hit_at_k,
            "recall_at_k": recall_at_k,
            "reciprocal_rank": reciprocal_rank
        })

    return results
```

```python

def hybrid_search(question, top_k=5):

    # 1. FAISS semantic search
    query_embedding = model.encode(
        [question],
        normalize_embeddings=True
    )

    query_vector = np.asarray(
        query_embedding,
        dtype="float32"
    )

    _, faiss_indices = index.search(
        query_vector,
        index.ntotal
    )

    faiss_ranking = [
        int(i) for i in faiss_indices[0] if i != -1
    ]

    # 2. BM25 keyword search
    query_tokens = question.lower().split()

    bm25_scores = bm25.get_scores(query_tokens)

    bm25_ranking = np.argsort(
        bm25_scores
    )[::-1].tolist()

    # 3. Combine rankings using RRF
    rrf_scores = {}

    for ranking in [faiss_ranking, bm25_ranking]:
        for rank, chunk_id in enumerate(ranking, start=1):
            rrf_scores[chunk_id] = (
                rrf_scores.get(chunk_id, 0)
                + 1 / (60 + rank)
            )

    # 4. Sort by combined score
    sorted_chunks = sorted(
        rrf_scores,
        key=rrf_scores.get,
        reverse=True
    )

    return sorted_chunks[:top_k]
```

```python

def section_aware_search(question, top_k=5):

    question_lower = question.lower()

    # Detect questions asking for overall findings
    if (
        "main findings" in question_lower
        or "conclusions" in question_lower
    ):
        # Conclusion section identified in our PDF
        return [77, 78, 79, 80][:top_k]

    # Otherwise, use our original FAISS search
    query_embedding = model.encode(
        [question],
        normalize_embeddings=True
    )

    query_vector = np.asarray(
        query_embedding,
        dtype="float32"
    )

    _, indices = index.search(query_vector, top_k)

    return [
        int(i) for i in indices[0] if i != -1
    ]


# Test Question 5
question = evaluation_questions[4]

retrieved = section_aware_search(question)

print("Retrieved:", retrieved)
print("Relevant:", ground_truth[question])
```

```python

import re

def find_conclusion_chunks(chunks):

    conclusion_ids = []
    inside_conclusion = False

    for i, chunk in enumerate(chunks):

        text = chunk["text"]

        # Detect the start of the Conclusion section
        if re.search(r"\b4\s+Conclusion\b", text, re.IGNORECASE):
            inside_conclusion = True

        # Detect the start of the Future Work section
        if re.search(r"\b5\s+Future Work\b", text, re.IGNORECASE):
            if inside_conclusion:
                conclusion_ids.append(i)
            break

        if inside_conclusion:
            conclusion_ids.append(i)

    return conclusion_ids


conclusion_ids = find_conclusion_chunks(chunks)

print("Automatically detected conclusion chunks:")
print(conclusion_ids)
```

```python

def improved_search(question, top_k=5):

    question_lower = question.lower()

    # Detect questions about the paper's conclusions
    conclusion_keywords = [
        "main findings",
        "conclusions",
        "key findings",
        "main results"
    ]

    if any(
        keyword in question_lower
        for keyword in conclusion_keywords
    ):
        # Find conclusion chunks automatically
        return find_conclusion_chunks(chunks)[:top_k]

    # Otherwise, use regular FAISS search
    query_embedding = model.encode(
        [question],
        normalize_embeddings=True
    )

    query_vector = np.asarray(
        query_embedding,
        dtype="float32"
    )

    _, indices = index.search(
        query_vector,
        min(top_k, index.ntotal)
    )

    return [
        int(i) for i in indices[0] if i != -1
    ]
```

```python

def evaluate_improved_retrieval(questions, ground_truth):

    results = []

    for question in questions:

        retrieved = improved_search(question, top_k=5)
        relevant = set(ground_truth[question])

        hits = [
            i for i in retrieved
            if i in relevant
        ]

        hit = int(len(hits) > 0)
        recall = len(hits) / len(relevant)

        rr = 0
        for rank, chunk_id in enumerate(retrieved, start=1):
            if chunk_id in relevant:
                rr = 1 / rank
                break

        results.append({
            "question": question,
            "hit": hit,
            "recall": recall,
            "rr": rr
        })

    return results


improved_results = evaluate_improved_retrieval(
    evaluation_questions,
    ground_truth
)

new_hit = np.mean([r["hit"] for r in improved_results])
new_recall = np.mean([r["recall"] for r in improved_results])
new_mrr = np.mean([r["rr"] for r in improved_results])

print("BASELINE VS IMPROVED RETRIEVAL")
print("------------------------------")
print(f"Hit@5:    0.800 -> {new_hit:.3f}")
print(f"Recall@5: 0.800 -> {new_recall:.3f}")
print(f"MRR@5:    0.700 -> {new_mrr:.3f}")
```

```python

def ask_improved_rag(question, top_k=5):

    # Retrieve evidence using our improved search
    retrieved_ids = improved_search(question, top_k=top_k)

    # Build context with page citations
    context = "\n\n".join(
        f"[Page {chunks[i]['page']}]\n{chunks[i]['text']}"
        for i in retrieved_ids
    )

    # Instruct Qwen to use only retrieved evidence
    messages = [
        {
            "role": "system",
            "content": (

"Answer the user's question using only the provided context. "
"Do not invent information. "
"IMPORTANT: Every factual statement must include a page citation "
"in the exact format [Page N]. "
"Place citations immediately after the claims they support. "
"Only cite page numbers appearing in the provided context. "
"If the evidence is insufficient, explicitly say so."

            )
        },
        {
            "role": "user",
            "content": (
                f"Context:\n{context}\n\n"
                f"Question: {question}"
            )
        }
    ]

    # Apply Qwen chat template
    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt",
        return_dict=True
    )

    inputs = {
        key: value.to(llm.device)
        for key, value in inputs.items()
    }

    # Generate an answer
    output = llm.generate(
        **inputs,
        max_new_tokens=250,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

    # Decode only newly generated tokens
    generated_tokens = output[0][
        inputs["input_ids"].shape[1]:
    ]

    answer = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return {
        "question": question,
        "answer": answer,
        "retrieved_ids": retrieved_ids,
        "source_pages": sorted(
            set(chunks[i]["page"] for i in retrieved_ids)
        )
    }


# Test our previously problematic question
result = ask_improved_rag(
    "What are the main findings of this study?"
)

print("ANSWER:")
print(result["answer"])

print("\nRETRIEVED CHUNKS:")
print(result["retrieved_ids"])

print("\nSOURCE PAGES:")
print(result["source_pages"])
```

```python

import re

def validate_citations(answer, source_pages):

    # Find all citations in the generated answer
    citations = re.findall(
        r"\[Page\s+(\d+)\]",
        answer
    )

    cited_pages = [int(page) for page in citations]

    # Identify citations pointing to unavailable pages
    invalid_pages = [
        page for page in cited_pages
        if page not in source_pages
    ]

    return {
        "has_citations": len(cited_pages) > 0,
        "cited_pages": cited_pages,
        "invalid_pages": invalid_pages,
        "citation_format_valid": (
            len(cited_pages) > 0
            and len(invalid_pages) == 0
        )
    }


# Validate our current answer
citation_report = validate_citations(
    result["answer"],
    result["source_pages"]
)

print(citation_report)
```

```python

def format_rag_response(result):

    answer = result["answer"]

    # Pages retrieved from the PDF
    source_pages = result["source_pages"]

    # Format the source list
    sources = ", ".join(
        f"Page {page}"
        for page in source_pages
    )

    # Build the final response
    formatted_response = (
        f"{answer}\n\n"
        f"---\n"
        f"Retrieved sources: {sources}"
    )

    return formatted_response


# Test with our existing answer
final_response = format_rag_response(result)

print(final_response)
```
